# Categorical Data Plots
Now let's discuss using seaborn to plot categoricala data! There are a few main plot types for this:

* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

Let's go through examples of each!

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot(df)
```

```python
sns.countplot(x = df["sex"], hue=df["smoker"])
```

```python
sns.barplot(x = df["sex"], y=df["total_bill"])
```
