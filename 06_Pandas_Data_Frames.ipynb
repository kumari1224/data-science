# Pandas DataFrame

A **DataFrame** in Pandas is a two-dimensional, tabular data structure organized into rows and columns, similar to an Excel spreadsheet or a SQL table.

Each column can contain a different data type, such as integers, strings, floats, or dates.

### Key Components
- **Rows:** Represent individual records.
- **Columns:** Represent attributes or features.
- **Index:** Labels used to identify rows (default: `0, 1, 2, ...`).
- **Column Names:** Labels used to identify each column.

### Example

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 28],
    "City": ["Delhi", "Mumbai", "Noida"]
}

df = pd.DataFrame(data)
print(df)
```

**Output:**
```text
      Name  Age    City
0    Alice   25   Delhi
1      Bob   30  Mumbai
2  Charlie   28   Noida
```

### Common Operations

- `df.head()` — View the first 5 rows.
- `df.shape` — Get the number of rows and columns.
- `df["Name"]` — Select a column.
- `df.iloc[0]` — Select the first row by position.
- `df["Age"].mean()` — Calculate the average age.

**In short:** A DataFrame is a labeled, two-dimensional data structure used to store, manipulate, filter, and analyze structured data in Python.

````python
data = [
    ["Alice", 25, "Delhi", 50000],
    ["Bob", 30, "Mumbai", 65000],
    ["Charlie", 28, "Noida", 55000],
    ["David", 35, "Bangalore", 80000],
    ["Emma", 24, "Delhi", 45000],
    ["Frank", 32, "Mumbai", 70000]
]

df_list_of_list_default_labels = pd.DataFrame(data)
df_list_of_list_default_labels
````

````python
df_list_of_list_with_labels = pd.DataFrame(data, columns=["Name", "Age", "City", "Salary"])
df_list_of_list_with_labels
````

**2. Selection and indexing of columns**

````python
df_list_of_list_with_labels['Name']
````

````python
df_list_of_list_with_labels[["Name", "Salary"]]
````

**3. Creating a new column**

````python
df_list_of_list_with_labels["Designation"] = ["Doctor", "Eng.", "Doctor", "HR", "Eng.", "Teacher"]
````

````python
df_list_of_list_with_labels
````

**4. Removing Columns**

````python
df_list_of_list_with_labels.drop("Designation", axis = 1)
````

````python
df_list_of_list_with_labels # Normal Drop won't remove the data by reference, for in-place you need to provide additional argument (see below)
````

````python
df_list_of_list_with_labels.drop("Designation", axis = 1, inplace = True)
df_list_of_list_with_labels
````

````python
# Adding the column back since remove by reference was done earlier using in-place arg
df_list_of_list_with_labels["Designation"] = ["Doctor", "Eng.", "Doctor", "HR", "Eng.", "Teacher"]
````

````python
df_list_of_list_with_labels
````

````python
df_list_of_list_with_labels.drop(0, axis = 0) # Remove first row now ( Alice will be gone in the returned copy)
````

**5. Selecting Rows**

````python
# df_list_of_list_with_labels[0] -> Wrong approach to select a row, it won't work in Data frames

df_list_of_list_with_labels.loc[0]
````

````python
df_list_of_list_with_labels.loc[[3,4]] # Select mulitiple rows
````

````python
df_list_of_list_with_labels.iloc[0] # Same as "df_list_of_list_with_labels.loc[0]" to some extent
````

**6. Selecting Subsets of Rows and Columns**

````python
df_list_of_list_with_labels.loc[[3,4]][["Name", "City", "Salary"]] # Select 3 and 4th row in which only select columns Name, City, Salary
````

**7. Conditional Selection**

````python
# Select only those people whose age > 30
df_list_of_list_with_labels[df_list_of_list_with_labels["Age"] > 30]
````

````python
# Select people age > 30 and they live in city = Mumbai
df_list_of_list_with_labels[(df_list_of_list_with_labels["Age"] > 30) & (df_list_of_list_with_labels["City"] == "Mumbai")]
````
