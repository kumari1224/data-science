**1. Which team won the most matches?**

```python
df["match_winner"].value_counts()
```

```python
match_wins = df["match_winner"].value_counts()
sns.barplot(x=match_wins.values, y=match_wins.index, palette="viridis")
plt.xlabel("Number of Matches")
plt.ylabel("Team")
plt.title("Number of Matches Won by Each Team")
```

**2. Toss decision trends**

```python
df.head()
```

```python
sns.countplot(x="toss_decision", data=df, palette="viridis")
plt.xlabel("Toss Decision")
plt.ylabel("Count")
plt.title("Toss Decision Distribution")
```

**3. Toss winner vs Match winner**

```python
df[df['toss_winner'] == df['match_winner']]['match_id'].count() # How many teams won the toss as well as won the match together
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
perc = (count * 100)/df.shape[0]
perc.round(2)
```
