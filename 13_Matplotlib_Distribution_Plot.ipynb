# Distribution Plots in Data Visualization

**Distribution plots are used to visualize the distribution of quantitative (numerical) data.**

They help us understand how data values are spread, where most observations are concentrated, whether the data is skewed, and whether there are any outliers.

### Types of Distribution Plots

Some commonly used Seaborn plots for visualizing data distributions are:

1. **Histogram (`histplot`)** – Displays the frequency distribution of numerical data by grouping values into bins.
2. **Joint Plot (`jointplot`)** – Visualizes the relationship between two numerical variables, along with their individual distributions.
3. **Pair Plot (`pairplot`)** – Shows pairwise relationships between multiple numerical variables, along with their individual distributions.
4. **Rug Plot (`rugplot`)** – Displays individual observations as small marks along an axis, helping us understand where data points are concentrated.

**Note:** `distplot()` was a Seaborn function used for distribution visualization, but it has been deprecated. Use `histplot()`, `displot()`, or `kdeplot()` instead.

### Types of Data

**1. Quantitative Data (Numerical Data)**

Data that can be expressed in numerical form and measured or counted.

Examples:
- Age: 25, 30, 45
- Salary: 40,000, 60,000
- Height: 165 cm, 175 cm
- Population: 1,000,000

Quantitative data can be further classified into:
- **Discrete:** Countable values, such as the number of students.
- **Continuous:** Measurable values, such as height, weight, and temperature.

**2. Qualitative Data (Categorical Data)**

Data that represents categories, labels, or characteristics rather than numerical measurements.

Examples:
- Gender: Male, Female
- Department: HR, IT, Finance
- Country: India, USA, Japan
- Product Category: Electronics, Clothing, Furniture

Qualitative data can be further classified into:
- **Nominal:** Categories without a natural order, such as country or gender.
- **Ordinal:** Categories with a meaningful order, such as Low, Medium, and High.

**Important:** Qualitative data can sometimes be represented using numbers (for example, 1 = Male and 2 = Female). However, these numbers are category labels, not quantitative measurements.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips') # Loading pre existing data
df
```

```python
df['size'].unique()
```

**Histogram/Displot**

```python
sns.displot(df['total_bill']) # Histogram is used to create bell curve etc
```

```python
sns.displot(df['total_bill'], bins=100) # Histogram is used to create bell curve etc
```

```python
sns.displot(df['total_bill'], bins=100, kde=True) # kde(kernal distribution estimator) -> Used to determine the skewness of the plot, like whether it is lef/right skewed
```

```python
plt.subplot(1, 2, 1)
sns.histplot(df['total_bill'], bins=20, kde = True)
plt.subplot(1, 2, 2)
sns.histplot(df['tip'], kde=True)
```

**JointPlot**

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='scatter')
```

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='hex')
```

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='kde')
```

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='reg')
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue='sex') # Hue -> group them on the basis of a particular category (ex: male, female etc)
```

```python
sns.pairplot(df, hue='total_bill', palette='rainbow')
```

```python
sns.rugplot(df['total_bill'])
```

```python
sns.rugplot(df)
```
