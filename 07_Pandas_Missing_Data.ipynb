```python
import numpy as np
import pandas as pd
```

**Finding Missing Data**

```python
data = {
    "A": ["Alice", np.nan, "Bob", "Charlie", np.nan],
    "B": [12, np.nan, 30, 28, 45],
    "C": [1, 2, 3, np.nan, np.nan],
    "D": [1, np.nan, np.nan, np.nan, 5],
    "E": [6,3,1,3,5]
}

df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any() # Column having at least one Null value
```

**Removing Missing Data**

```python
df.dropna() # All the rows has Null value in it so the result will be empty
```

```python
df
```

```python
df.dropna(thresh=3) # Row at least should have 3-non Null values
```

**Filling the Missing Data**

```python
df.fillna(0) # Mutation is not in-place ( will have to provider separate inplace argument in the fillna() call)
```

```python
default_null_replacement = { 'A': 'John Doe', 'B': 100, 'C': 20, 'D': 400}
df.fillna(value = default_null_replacement)
```

```python
df
```

```python
df.fillna(df[["B", "C", "D", "E"]].mean()) # Fill the null values with the "Mean" since column "A" has
                                           # string values so we slice that and then find the mean
```
