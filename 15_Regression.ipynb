# Implot()
Regression plot is used in machine learning, in this notebook only overview will be discussed

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
```

```python
sns.lmplot(x="total_bill", y="tip", data=tips)
plt.show()
```

```python
sns.lmplot(x="total_bill", y="tip", data=tips, hue="sex")
plt.show()
```

```python
sns.lmplot(x="total_bill", y="tip", data=tips, hue="sex", palette="rainbow")
plt.show()
```

```python
sns.lmplot(x="total_bill", y="tip", data=tips, hue="sex", palette="coolwarm", markers=['o', 'v'])
plt.show()
```

```python
sns.lmplot(x="total_bill", y="tip", data=tips, hue="sex", palette="coolwarm", markers=['o', 'v'], scatter_kws={'s': 100})
plt.show()
```
