```python
import re

def extract_episodes(txt):
  episodes = re.search(r"Episodes: \s*(\d+)", txt)
  return int(episodes.group(1)) if episodes else None
```

```python
df["Episodes"] = df["Title"].apply(extract_episodes)
df
```

**Make a new column for Run time**

```python
def extract_time_stamp(txt):
  time_stamp = re.search(r"Runtime: \s*(\d+) \s*min", txt)
  return int(time_stamp.group(1)) if time_stamp else None
```

```python
df["Timestamp"] = df["Title"].apply(extract_time_stamp)
df
```
