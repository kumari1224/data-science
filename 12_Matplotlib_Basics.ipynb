```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x = np.linspace(0, 5, 11)
y = x ** 2
print("X =>", x)
print("Y =>", y)
```

```python
plt.plot(x, y)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Title')
plt.show()
```

**Subplots in Matlplotlib**


```python
plt.subplot(2,2,1) # Suplots will be drawn for rows and columns format ( )
plt.plot(x**3,y**3) # Will will create x,y plot in the 1st plot
plt.subplot(2,2,4)
plt.plot(x,y) # Will will create x,y plot in the 4th plot
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2, y**2)
```

```python
plt.subplot(6,6,1)
plt.subplot(6,6,2)
plt.subplot(6,6,3)
plt.subplot(6,6,4)
plt.subplot(6,6,5)
plt.subplot(6,6,6)
plt.subplot(6,6,7)
plt.subplot(6,6,8)
plt.subplot(6,6,9)
```

**Object Oriented Way**

```python
fig = plt.figure() # Creating a blank canvas first
axis1 = fig.add_axes([0.1, 0.1, 1, 1]) # [left, bottom, width, height]
axis1.plot(x,y)

axis2 = fig.add_axes([0.2, 0.2, 0.3, 0.3]) # You will see the second graph will be overlapping the first
axis2.plot(x,y)
```

```python
fig = plt.figure(figsize=(20,8), dpi=500) # Canvas size can be controlled too, dpi -> the quality and clarity of the graph
axis1 = fig.add_axes([0.1, 0.1, 1, 1]) # [left, bottom, width, height]
axis1.plot(x,y)

axis2 = fig.add_axes([0.2, 0.2, 0.3, 0.3]) # You will see the second graph will be overlapping the first
axis2.plot(x,y)
```

```python

# Create a figure
fig, axes = plt.subplots(4, 3, figsize=(18, 19))
fig.suptitle("Matplotlib Visualization Showcase",
             fontsize=22, fontweight="bold")

# Generate sample data
x = np.linspace(0, 10, 200)
y = np.sin(x)

# 1. Line plot
axes[0, 0].plot(x, y, color="royalblue", linewidth=2)
axes[0, 0].set_title("1. Line Plot")

# 2. Bar chart
categories = ["A", "B", "C", "D"]
values = [25, 40, 30, 55]
axes[0, 1].bar(categories, values,
               color=["tomato", "orange", "seagreen", "royalblue"])
axes[0, 1].set_title("2. Bar Chart")

# 3. Scatter plot
axes[0, 2].scatter(np.random.rand(100),
                   np.random.rand(100),
                   c=np.random.rand(100),
                   cmap="plasma", s=60)
axes[0, 2].set_title("3. Scatter Plot")

# 4. Pie chart
axes[1, 0].pie([35, 25, 20, 20],
               labels=["A", "B", "C", "D"],
               autopct="%1.1f%%",
               colors=["#ff9999", "#66b3ff", "#99ff99", "#ffcc99"])
axes[1, 0].set_title("4. Pie Chart")

# 5. Histogram
axes[1, 1].hist(np.random.randn(1000),
                bins=30, color="mediumpurple", edgecolor="black")
axes[1, 1].set_title("5. Histogram")

# 6. Area plot
axes[1, 2].fill_between(x, y, color="skyblue", alpha=0.6)
axes[1, 2].plot(x, y, color="navy")
axes[1, 2].set_title("6. Area Plot")

# 7. Heatmap
heatmap = np.random.rand(10, 10)
axes[2, 0].imshow(heatmap, cmap="inferno")
axes[2, 0].set_title("7. Heatmap")

# 8. Box plot
axes[2, 1].boxplot([np.random.randn(100),
                    np.random.randn(100) + 2,
                    np.random.randn(100) + 4])
axes[2, 1].set_title("8. Box Plot")

# 9. Polar plot
theta = np.linspace(0, 2*np.pi, 300)
r = 2 + np.sin(6 * theta)
axes[2, 2].remove()
ax = fig.add_subplot(4, 3, 9, projection="polar")
ax.plot(theta, r, color="crimson")
ax.set_title("9. Polar Plot")

# 10. Contour plot
X, Y = np.meshgrid(np.linspace(-3, 3, 100),
                   np.linspace(-3, 3, 100))
Z = np.sin(X**2 + Y**2)
axes[3, 0].contourf(X, Y, Z, levels=30, cmap="coolwarm")
axes[3, 0].set_title("10. Contour Plot")

# 11. Bubble chart
sizes = np.random.randint(50, 1000, 50)
axes[3, 1].scatter(np.random.rand(50),
                   np.random.rand(50),
                   s=sizes, c=sizes,
                   cmap="viridis", alpha=0.6)
axes[3, 1].set_title("11. Bubble Chart")

# 12. Generative art: colorful spiral
t = np.linspace(0, 20*np.pi, 2000)
r = t / 10
axes[3, 2].scatter(r*np.cos(t), r*np.sin(t),
                   c=t, cmap="turbo", s=3)
axes[3, 2].set_title("12. Generative Art")

plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
