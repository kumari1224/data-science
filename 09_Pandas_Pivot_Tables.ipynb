```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010, 1011, 1012],
    "Order_Date": [
        "2026-01-05", "2026-01-08", "2026-01-12", "2026-01-15",
        "2026-02-02", "2026-02-10", "2026-02-14", "2026-02-20",
        "2026-03-03", "2026-03-09", "2026-03-15", "2026-03-22"
    ],
    "Employee": [
        "Alice", "Bob", "Alice", "Charlie", "Bob", "Alice",
        "Charlie", "Bob", "Alice", "Charlie", "Bob", "Alice"
    ],
    "Department": [
        "Electronics", "Clothing", "Furniture", "Electronics",
        "Clothing", "Furniture", "Electronics", "Clothing",
        "Furniture", "Electronics", "Clothing", "Furniture"
    ],
    "Region": [
        "North", "South", "East", "West", "North", "South",
        "East", "West", "North", "South", "East", "West"
    ],
    "Quantity": [2, 5, 1, 3, 4, 2, 6, 3, 2, 4, 5, 1],
    "Unit_Price": [15000, 800, 12000, 20000, 1000, 10000, 18000, 900, 14000, 22000, 1200, 11000],
    "Payment_Mode": [
        "Card", "Cash", "UPI", "Card", "UPI", "Cash",
        "Card", "UPI", "Cash", "Card", "UPI", "Cash"
    ],
    "Customer_Rating": [4, 5, 3, 4, 5, 4, 2, 3, 5, 4, 3, 5]
}

df = pd.DataFrame(data)
df
```

```python
df['Month'] = pd.to_datetime(df['Order_Date']).dt.month_name() # Converting to Order_Date values to datetime like because they are strings in the input
df
```

```python
df['Quarter'] = 'Q' + pd.to_datetime(df['Order_Date']).dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df, values="Unit_Price", index="Region", columns="Department") # Aggregation func Default is "mean"
```

```python
pd.pivot_table(df, values="Unit_Price", index="Region", columns="Department", aggfunc="median")
```

```python
pd.pivot_table(df, values=["Unit_Price", "Quantity"], index="Region", columns="Department")
```
