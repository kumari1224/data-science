```python
import numpy as np
import pandas as pd
```

**Merging 2 dataframes**

```python
employees = pd.DataFrame({
    "employee_id": [101, 102, 103, 104, 106],
    "name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "department": ["Engineering", "HR", "Finance", "Engineering", "Marketing"]
})

# DataFrame 2: Salary info
salaries = pd.DataFrame({
    "employee_id": [101, 102, 103, 104, 107],
    "salary": [80000, 60000, 75000, 90000, 65000],
    "bonus": [10000, 5000, 8000, 12000, 6000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees, salaries) # Merge two Data Frames
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "inner") # Common ids will be included only and rest will be ignored
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "outer") # Everything will be included and empty cell will be filled with NaN
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "left") # Data will be merged on the basis of Employees
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "right") # Data will be merged on the basis of Salaries
```

**Concatenation of 2 dataframes**

```python
dict1 = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "age": [25, 30, 28, 35, 24]
})

dict2 = pd.DataFrame({
    "id": [6, 7, 8, 9, 10],
    "name": ["Frank", "Grace", "Henry", "Ivy", "Jack"],
    "age": [32, 27, 29, 31, 26]
})
```

```python
dict1
```

```python
dict2
```

```python
pd.concat([dict1, dict2]) # Data concatenating on the basis of column
```

```python
pd.concat([dict1, dict2], axis=1) # Data concatenating on the basis of row
```

**Joining 2 Data frames**

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1,2,3])

df2 = pd.DataFrame({
    'score': [67, 90, 56]
}, index=[2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df2.join(df1)
```

```python
df1.join(df2, how="outer")
```

```python
df2.join(df1, how="outer")
```
