## Pandas Series

A Series is a 1-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

### Creating Series
There are multiple ways to create a Pandas Series



```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
my_list = [10, 20, 30]
arr = np.array([40, 50, 60])
dictionary = { 'd': 11, 'e': 22, 'f': 33 }
```

```python
pd.Series(my_list) # Indexes will be the numbers starting with 0
```

```python
pd.Series(my_list, index = labels) # Set indices/labels/keys instead of 0,1,2,... etc
```

```python
pd.Series(arr) # 1-D Array is fine but 2-D array will throw error because Series itself is 1D
```

```python
pd.Series(dictionary) # Dictionary will have labels/index/keys as keys and value will be the value
```
