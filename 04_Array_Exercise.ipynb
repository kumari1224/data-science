**Question 1: Valid Sudoku**

```python
# Example of a valid array
s1 = [
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
]

# Example of an invalid array
s2 = [
    [5, 3, 5, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
]
```

```python
def get_cell_sum(matrix, axis = 0):
  # get the sum of the rows whtne axis = 1 else column
  return np.sum(matrix, axis)

def is_sudoku_valid(matrix):
  # Convert the array into numpy array first
  arr = np.array(matrix)

  # check if the rows sum == 45
  r = get_cell_sum(arr, 1)
  for i in r:
    if i != 45:
      print("Sudoku is not valid [Err=1]")
      return

  # check if the cols sum == 45
  c = get_cell_sum(arr, 0)
  for i in c:
    if i != 45:
      print("Sudoku is not valid [Err=2]")
      return

  # check if the sum of 3x3 grid == 45
  # arr.slice(0:3, 0:3), arr.slice(0:3, 3:6) ..
  for i in range(0, 9, 3):
    for j in range(0, 9, 3):
      n = arr[i:i+3, j:j+3]
      sum = n.sum()
      if sum != 45:
        print("Sudoku is not valid [Err=3]")
        return

  print("SUDOKU IS VALID HURRAY !!!!")
```

```python
is_sudoku_valid(s1)
```

```python

is_sudoku_valid(s2)
```
