```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print(A)
print(B)
```

```python
A @ B # Dot product of 2 Matrix: [[a, b], [c, d]] * [[e, f], [g, h]] => [[ae + bg, af + bh], [ce + dg, cf + dh]]
```

```python
np.dot(A,B) # Alternative to A @ B
```

```python
A.T # Transform Matrix Operation
```

```python
a = np.array([[1,2,3,4,5]])
b = np.array([[6,7,8,9,0]])

# Stacking Array -> Vertical, Horizontal

# Vertical Format Stacking
print("Vertical Stacking :", np.vstack((a,b))) # Tuple format input

# Vertical Format Stacking
print("Horizontal Stacking :", np.hstack((a,b))) # Tuple format input
```

```python
np.column_stack((a,b)) # Column stack
```

```python
# Split -> Vertical, Horizontal
c = np.arange(16).reshape(4,4)
print("Original Array :\n", c)

# Horizontal Division
print("Horizontal Division :\n", np.hsplit(c, 2))

# Vertical Division
print("Vertical Division :\n", np.vsplit(c, 2))
```
